# Exploratory Analysis

This notebook provides shot maps, xG flow, passing networks, and tactical insights using StatsBomb Open Data.

```python
import pandas as pd
import matplotlib.pyplot as plt
from mplsoccer import Pitch

events = pd.read_csv('../data/processed/events.csv')
shots = events[events['event_type'] == 'Shot'].copy()
```

## Shot Map

```python
pitch = Pitch(pitch_type='statsbomb', line_color='black')
fig, ax = pitch.draw(figsize=(10, 7))
pitch.scatter(shots['x'], shots['y'], s=20, c=shots['shot_statsbomb_xg'],
              cmap='Reds', ax=ax, alpha=0.7)
plt.title('Shot Map (color = StatsBomb xG)')
plt.show()
```

## xG Flow Chart

```python
xg_flow = shots.groupby('minute')['shot_statsbomb_xg'].sum()
xg_flow.plot(figsize=(10, 4), color='tomato')
plt.title('xG Flow by Minute')
plt.ylabel('xG')
plt.show()
```
